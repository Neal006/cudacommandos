```python
import pandas as pd
```

```python
s1 = pd.read_csv("train_source1.tsv", sep="\t")
s2 = pd.read_csv("train_source2.tsv", sep="\t")
s3 = pd.read_csv("train_source3.tsv", sep="\t")
gt = pd.read_csv("train_ground_truth.tsv", sep="\t")

print(s1.shape, s2.shape, s3.shape, gt.shape)
print(s1.head())
print(gt.head(10))

# singleton fraction
singleton_frac = gt["matched_entity_ids"].isna().mean() + (gt["matched_entity_ids"] == "").mean()
print("singleton fraction:", singleton_frac)
```

```python
labeled_ids = set(gt["source1_entity_id"])
all_s1_ids = set(s1["entity_id"])
print("S1 total:", len(all_s1_ids))
print("S1 labeled:", len(labeled_ids))
print("S1 unlabeled:", len(all_s1_ids - labeled_ids))
```

```python
row = gt.iloc[0]
s1_id = row["source1_entity_id"]
matched_ids = row["matched_entity_ids"].split(",")

print("S1 record:")
print(s1[s1["entity_id"] == s1_id][["business_name", "business_address", "country"]])

print("\nMatched records:")
for mid in matched_ids:
    src = s2 if mid.startswith("S2-") else s3
    print(src[src["entity_id"] == mid][["entity_id", "business_name", "business_address", "country"]])
```

```python
print(gt["source1_entity_id"].head(3).tolist())
print(s1["entity_id"].head(3).tolist())
```

```python
print(len(labeled_ids & all_s1_ids))  # actual overlap
```

```python
import re

def digit_len(id_series):
    return id_series.str.extract(r'S1-(\d+)')[0].str.len().value_counts()

print("gt id lengths:")
print(digit_len(gt["source1_entity_id"]))

print("\ns1 id lengths:")
print(digit_len(s1["entity_id"]))
```
